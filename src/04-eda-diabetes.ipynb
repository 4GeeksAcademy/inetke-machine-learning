## Paso 1: Recopilación de datos y definición del problema

Este conjunto de datos proviene originalmente del Instituto Nacional de Diabetes y Enfermedades Digestivas y Renales. El objetivo es predecir en base a medidas diagnósticas si un paciente tiene o no diabetes.

> A partir de las variables clínicas, predecir si el paciente es diabético (sí/no). Nuestro variable objetivo (y) es `Outcome`. 

```python
df = pd.read_csv('../data/raw/diabetes.csv')
df.head()
```

- `Pregnancies` Número de embarazos del paciente (numérico)
- `Glucose` Concentración de glucosa en plasma a las 2 horas de un test de tolerancia oral a la glucosa (numérico)
- `BloodPressure` Presión arterial diastólica (medida en mm Hg) (numérico)
- `SkinThickness` Grosor del pliegue cutáneo del tríceps (medida en mm) (numérico)
- `Insulin` Insulina sérica de 2 horas (medida en mu U/ml) (numérico)
- `BMI` Índice de masa corporal (numérico)
- `DiabetesPedigreeFunction` Función de pedigrí de diabetes (numérico)
- `Age` Edad del paciente (numérico)
- `Outcome` Variable de clase (0 o 1), siendo 0 negativo en diabetes y 1, positivo (numérico)

```python
glucose_stats = df['Glucose'].describe()
glucose_stats
```

>Un valor de glucosa en plasma **cercano a 0 mg/dL es imposible** en una persona viva. Los valores pueden ser bastante altos en casos de diabetes no controlada. En un test de tolerancia, podrían superar los 300-400 mg/dL, e incluso más en casos extremos. En este caso decidimos dejar los valores altos y ahora veremos que hacemos con los valores de 0. 

```python
pressure_stats = df['BloodPressure'].describe()
pressure_stats
```

> Un valor por debajo de 30–40 mmHg en una persona consciente es muy raro y un **valor 0 mmHg es imposible**, casi seguro significa dato faltante o error de registro. 

```python
diabetes_function_stats = df['DiabetesPedigreeFunction'].describe()
diabetes_function_stats
```

>`DiabetesPedigreeFunction` no es una medida clínica directa como la glucosa, es un índice calculado que resume el riesgo genético/familiar de diabetes a partir de los antecedentes en la familia, entonces estos valores pueden ser reales.

```python
skin_thickness_stats = df['SkinThickness'].describe()
skin_thickness_stats
```

Un **valor 0 mm no puede realista**, será dato no medido (faltante), valores reales mínimos razonables serían ≈ 3–5 mm.

### Split (dos métodos o enfoques)

```python
X = df.drop('Outcome', axis=1)
y = df['Outcome']

# Instance of: from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,
                                                    y,
                                                    test_size=0.2,
                                                    random_state=18)
```

```python
X_train.to_csv('/workspaces/inetke-machine-learning/data/processed/X_train_diabetes', index=False)
X_test.to_csv('/workspaces/inetke-machine-learning/data/processed/X_test_diabetes', index=False)
```

## Paso 3: Construir un modelo de árbol de decisión

```python
df = X
df['Outcome'] = y

pd.plotting.parallel_coordinates(df, 'Outcome', color = ("#E58139", "#8139E5"))
plt.xticks(rotation=45)
```

```python
model = DecisionTreeClassifier(random_state = 42)
model.fit(X_train, y_train)
```

```python
fig = plt.figure(figsize=(15,15))

tree.plot_tree(model, feature_names = list(X_train.columns), class_names = ['0', '1'], filled = True)

plt.show()
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
accuracy_score(y_test, y_pred)
```
